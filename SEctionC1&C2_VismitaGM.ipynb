```python
#Swapping of Two no.s
a=int(input("Enter a:"))
b=int(input("Enter b:"))
#Using arithmetic way
a=a+b
b=a-b
a=a-b
print("a=",a)
print("b=",b)
#Using XOR
a=a^b
b=a^b
a=a^b
print("XOR:",a,b)
#Using tuple unpacking--->easiest way and most pythonic way
a,b=b,a
print("a=",a)
print("b=",b)
#tuple unpacking:python grps the values on the right into a tuble and unpacks them into variables on left
```

```python
#Swiggy order
price=float(input("Enter the price:"))
quantity=int(input("Enter no. of quantity:"))
subtotal=price*quantity
gst=(subtotal*18)/100
if subtotal<500:
    delivery=40
else :
    delivery=0
total=subtotal+gst+delivery
print("Subtotal:",subtotal)
print("GST(18%):",gst)
print("Delivery:",delivery)
print("------------------------")
print("Total:",total)
```

```python
#Coverting
temp=float(input("Enter temperature:"))
unit=input("Enter unit(C/F):").upper()
if unit=="C":
  result=(temp*9/5)+32
  print("Temperature in F:",round(result,2))
elif unit=="F":
  result=(temp - 32)*5/9
  print("Temperature in C:",round(result,2))
else:
  print("Invalid unit!!")
```

```python
n=int(input("Enter a 4-digit number:"))
if 1000<=n<=9999:
  d1=n%10
  d2=(n//10)%10
  d3=(n//100)%10
  d4=(n//1000)%10
  total=d1+d2+d3+d4
  print("Sum of digits in these are:",total)
else:
  print("Enter a 4-digit number!!")
```

```python
#Smart parser
n=input("Enter a value:")
try:
  n=int(n)
  datatype="int"
except ValueError:
  try:
    n=float(n)
    datatype="float"
  except ValueError:
    datatype="str"
print("Value:",n)
print("Dataype:",datatype)
```

```python
income=float(input("Enter your annual income:"))
tax=0
if income>300000:
  tax+=min(income-300000,400000)*0.05
if income>700000:
  tax+=min(income-700000,300000)*0.10
if income>100000:
  tax+=min(income-100000,200000)*0.15
if income>120000:
  tax+=min(income-120000,300000)*0.20
if income>150000:
  tax+=(income-150000)*0.30
print("Annual income:",income)
print("Total income tax:",tax)
```
