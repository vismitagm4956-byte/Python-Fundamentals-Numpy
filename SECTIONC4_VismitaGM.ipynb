```python
#31
count =0
print("Armstrong Numbers:")
for num in range(1,10000):
  temp=num
  digits=len(str(num))#length can be counted only in string format
  total=0
  while temp>0:
    digit=temp%10#getting last digit
    total+=digit**digits
    temp=temp//10
  if total==num:
    print(num,end=" ")
    count+=1
print("\nTotal count:",count)
```

```python
#32 Fib series
n=int(input("Enter N:"))
a=0
b=1
total=0
for i in range(n):
  print(a,end=" ")

  total+=a
  a,b=b,b+a
#Sum of fib series
print("\nSum=",total)
```

```python
#33 Check prime
n=int(input("Enter n:"))
total=0
for num in range(2,n+1):
  is_prime=True
  for i in range(2,int(num**0.5)+1):
    if num%i==0:
      is_prime=False
      break
  if is_prime:
    print(num,end=" ")
    total+=num
print("\nSum of prime numbers=",total)
```

```python
#34 Check palindrome
n=int(input("Enter a number:"))
orignal=n
reverse=0
if n<0:
  print("Not a palindrome!")
else:
  while n>0:
    digit=n%10
    reverse=reverse*10+digit
    n=n//10
  print("Reversed order:",reverse)
  if orignal==reverse:
    print("is a palindrome")
  else:
    print("Not a palindrome!")
```

```python
#35
n1=int(input("Enter 1st digit:"))
n2=int(input("Enter 2nd digit:"))
x=n1
y=n2
while y!=0:
  remainder=x%y
  x=y
  y=remainder
hcf=x
lcm=(n1*n2)//hcf
print("HCF:",hcf)
print("LCM=",lcm)
```

```python
#36
s=input("Enter a sentence:")
characters=0
words=0
vowel=0
consonants=0
digits=0
spaces=0
special=0
in_word=False
for ch in s:
  characters+=1
  if ch.isspace():
    spaces+=1
    in_word=True
  else:
    if not in_word:
      words==1
    in_word=True
    if ch.isalpha():
      if ch.lower() in "aeiou":
        vowel+=1
      else:
        consonants+=1
    elif ch.isdigit():
      digits+=1
    else:
      special+=1
print("Total Characters:",characters)
print("Total words:",words)
print("Vowels:",vowel)
print("Consonants:",consonants)
print("Digits:",digits)
print("Spaces:",spaces)
print("Special characters:",special)
```
